```python
# File Name: FruitVeggieMatcher.py
# Course: SDEV220
# Author: Cody Newkirk

# This program checks if something is small and/or green,
# and matches it to a cherry, pea, watermelon, or pumpkin.

# Assign True or False to each characteristic
small = True
green = True

# Check combinations and print the matching item
if small and green:
    print("pea")
elif small and not green:
    print("cherry")
elif not small and green:
    print("watermelon")
else:
    print("pumpkin")
```
