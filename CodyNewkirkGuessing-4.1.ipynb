```python
# File Name: Guessing.ipynb
# Course: SDEV220
# Author: Cody Newkirk

# This program compares a guessed number to a secret number
# and prints whether the guess is too low, too high, or just right.

# Choose a number between 1 and 10 and assign it to secret
secret = 7

# Choose another number between 1 and 10 and assign it to guess
guess = 5

# Check the guess
if guess < secret:
    print("too low")
elif guess > secret:
    print("too high")
else:
    print("just right")
```
