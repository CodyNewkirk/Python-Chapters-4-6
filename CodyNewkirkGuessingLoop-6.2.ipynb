```python
# File Name: GuessingLoop.ipynb
# Course: SDEV220
# Author: Cody Newkirk

# This program uses a while loop to compare 'number' with 'guess_me'

guess_me = 7
number = 1

while True:
    if number < guess_me:
        print("too low")
    elif number == guess_me:
        print("found it!")
        break
    else:
        print("oops")
        break
    number += 1
```
